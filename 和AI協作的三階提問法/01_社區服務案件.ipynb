> 課堂合成資料；一列代表 **一筆社區服務案件**。

| ① 看懂資料 | ② 一題一圖 | ③ 解讀數據與圖表 |
| :--- | :--- | :--- |
| 欄位與邊界 | 人機協作選圖與繪圖 | 核心結論・數據洞察・說出限制 |

**你負責提問與驗收，AI 負責寫程式。** 執行第一格載入資料與字型後，依序將三階 prompt 貼給 Colab Gemini 或能讀取筆記本的 AI，所有程式碼、數字表、圖表與解讀完全由 AI 產出。

**範例問題：**2025 年已結案案件中，哪個服務類別的處理時數中位數最高？（可在第一階自訂或由 AI 提出 5 個問題）　｜　[下載資料集](./資料集/%E7%A4%BE%E5%8D%80%E6%9C%8D%E5%8B%99%E6%A1%88%E4%BB%B6.xlsx)

````python
# 0｜載入 xlsx，並沿用前面章節的繁體中文字型設定
from pathlib import Path
from urllib.request import urlretrieve
import glob
import os
import pandas as pd
import matplotlib.pyplot as plt

filename = '社區服務案件.xlsx'
file_path = next((p for p in [Path('資料集') / filename, Path(filename), Path('和AI協作的三階提問法/資料集') / filename] if p.exists()), None)
if file_path is None:
    file_path = Path(filename)
    urlretrieve('https://raw.githubusercontent.com/roberthsu2003/ai-excel-analytics/main/%E5%92%8CAI%E5%8D%94%E4%BD%9C%E7%9A%84%E4%B8%89%E9%9A%8E%E6%8F%90%E5%95%8F%E6%B3%95/%E8%B3%87%E6%96%99%E9%9B%86/%E7%A4%BE%E5%8D%80%E6%9C%8D%E5%8B%99%E6%A1%88%E4%BB%B6.xlsx', file_path)
%pip install -q mplfonts
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font
for font_file in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), 'fonts', '*.[ot]tf')):
    font_manager.fontManager.addfont(font_file)
use_font('Noto Sans CJK SC')
plt.rcParams['axes.unicode_minus'] = False
df = pd.read_excel(file_path, sheet_name='資料')
print('資料檔：', file_path, '；列數：', len(df))
display(df.head())
````

---

## ① 第一階｜看懂資料

**先自己判斷：**處理時數＝指標；受理日期、服務類別、案件狀態＝條件；案件編號＝標識。處理中的缺值不能當成 0 小時。

**複製給 AI**

```text
請讀取已載入的 df（社區服務案件.xlsx／工作表「資料」）：
1. 顯示前 5 列、總列數、欄位名稱、每欄缺值數
2. 說明一列代表什麼，將欄位分成指標、條件、標識
3. 檢查資料的時間範圍、重複或缺值
4. 提出 5 個可以回答的問題（涵蓋不同分析面向），以及 1 個不能回答的問題並說明理由
⚠️ 先不要畫圖，也不要下結論。
```

**驗收：**筆數、缺值與欄位角色，都要和 xlsx 一致。

---

## ② 第二階｜一個問題、一張圖（人機協作選圖）

> 🎯 **人機協作流程：**
> 1. 第一階 AI 會提出 5 個可以回答的問題，你可以從中挑選一個想探索的問題（也可以重複第 2 階與第 3 階練習 5 次！）。
> 2. **先請 AI 建議圖表並分析優缺點**，先不要急著畫圖。
> 3. 你學過第 6 課的選圖規則，由你評估後**決定最終圖表**，再請 AI 出表與畫圖。

### 步驟 2-1｜填入問題，請 AI 建議適合圖表與優缺點

**複製給同一段 AI 對話（將問題替換為你想分析的問題）：**

```text
延續第一階，我想針對這個問題進行深入分析：
『＿＿＿＿＿＿＿＿＿＿＿＿』
（範例問題：2025 年已結案案件中，哪個服務類別的處理時數中位數最高？）

請先不要寫程式畫圖，請先：
1. 說明這個問題屬於哪種問題類型（比較、趨勢、占比、分布、關聯）。
2. 建議 2～3 種適合呈現此問題的圖表類型，並條列分析每種圖表的「優點」與「缺點」（請考慮類別數量、名稱長短、極端值等特性）。
3. 說明你最推薦哪一種圖表，以及理由是什麼。
等我確認要採用哪種圖表後，我們再進行下一步。
```

---

### 步驟 2-2｜決定圖表，請 AI 產出數字表與繪圖

**看過 AI 的分析後，運用第 6 課選圖思考規則決定圖表，再貼出以下 Prompt：**

```text
我決定採用【＿＿圖】（例如：水平長條圖）。

請先產出排序數字表，再只畫一張【＿＿圖】。
繪圖注意事項：
1. 排除無效或缺值的資料，數值軸請從 0 開始
2. 圖上清楚標註數值與單位，文字使用繁體中文（沿用已設定的中文字型）
⚠️ 一次只回答這個問題，不要畫其他圖。
```

> 💡 **官方範例情境參考：** 若詢問「各服務類別處理時數中位數」，因為有 8 個類別且名稱較長，【水平長條圖】文字不易重疊且容易比較高低，是最佳選擇。

**驗收：**先核對數字表，再核對圖表類型、坐標軸、單位與中文字。

---

## ③ 第三階｜解讀數據與圖表（說出洞察與限制）

沿用第二階的數字表與圖表，留在同一段 AI 對話中，請 AI 解釋分析結果。

**複製給同一段 AI 對話：**

```text
請根據剛才產出的數字表與圖表，說明如何正確解釋這些分析結果。

解讀注意事項：
1. 核心結論：回答剛才設定的問題，並引用圖表中的具體數據支持
2. 數據洞察：說明數據與圖表呈現的重要現象或趨勢特徵
3. 資料限制：寫明統計範圍、分母，以及是否有被排除的資料（如缺值）
4. 不能推論的事：說明單憑目前數據無法證明什麼（不把相關當因果、不妄下過度結論）
```

**你要檢查（驗收標準）：**
- [ ] 結論有具體數據支持，與圖表完全一致
- [ ] 百分比或比率有清楚標明分母
- [ ] 沒有把相關說成因果，明確指出資料限制
